### 4.2. Representación de un problema mediante estados
#### Parte A: Completar la cha del problema para el siguiente caso: un repartidor debe ir desde el Almacén hasta la Zona Norte, pudiendo pasar por distintos puntos intermedios de la ciudad.r

| Elemento | Descripción |
| :--- | :--- |
| **Estado inicial** | Repartidor en el Almacén. |
| **Estado objetivo** | Repartidor en la Zona Norte. |
| **Acciones posibles** | Desplazarse hacia un punto directamente conectado con la ubicación actual (moverse a un nodo vecino). |
| **Costo de cada acción** | 1 paso (costo unitario por desplazamiento). |

#### Parte B: Representar el mapa de la ciudad como un grafo en Python, usando un diccionario de listas (cada clave es un punto, cada valor es la lista de puntos conectados directamente):

```python
mapa = {
    "Almacen": ["Centro", "Terminal"],
    "Centro": ["Almacen", "ZonaNorte", "Mercado"],
    "Terminal": ["Almacen", "Mercado"],
    "Mercado": ["Centro", "Terminal", "ZonaNorte"],
    "ZonaNorte": ["Centro", "Mercado"],
}

print(mapa)
```

#### 4.3. Implementación de BFS (Búsqueda en Amplitud)

```python
def bfs(grafo, inicio, objetivo):
    """
    Busca el camino mas corto entre el inicio y el objetivo usando BFS.
    """

    cola = [[inicio]]  # Lista que funciona como cola y guarda los caminos posibles
    visitados = set()  # set guarda nodos visitados sin repetirlos

    while cola:  # Se repite mientras existan caminos por revisar
        camino = cola.pop(0)  # Saca el primer camino de la cola, por eso es FIFO
        nodo_actual = camino[-1]  # Obtiene el ultimo nodo del camino actual

        if nodo_actual == objetivo:  # Verifica si ya se llego al destino
            return camino  # Devuelve el camino encontrado

        if nodo_actual not in visitados:  # Evita revisar un nodo que ya fue visitado
            visitados.add(nodo_actual)  # Agrega el nodo actual al conjunto de visitados

            for vecino in grafo[nodo_actual]:  # Recorre los puntos conectados al nodo actual
                nuevo_camino = camino + [vecino]  # Crea otro camino agregando el vecino
                cola.append(nuevo_camino)  # Agrega el nuevo camino al final de la cola

    return None  # Se devuelve None si no existe un camino


resultado = bfs(mapa, "Almacen", "ZonaNorte")  # Busca una ruta desde Almacen hasta ZonaNorte
print("Camino encontrado:", resultado)  # Muestra la ruta encontrada
```

| **Acciones posibles** | Desplazarse hacia un punto directamente conectado con la ubicación actual (moverse a un nodo vecino). |

**Parte B:**  Representar el mapa de la ciudad como un grafo en Python, usando un diccionario de listas (cada clave es un punto, cada valor es la lista de puntos conectados directamente):

```python
mapa = {
    "Almacen": ["Centro", "Terminal"],
    "Centro": ["Almacen", "ZonaNorte", "Mercado"],
    "Terminal": ["Almacen", "Mercado"],
    "Mercado": ["Centro", "Terminal", "ZonaNorte"],
    "ZonaNorte": ["Centro", "Mercado"],
}

print(mapa)
```

#### 4.3. Implementación de BFS (Búsqueda en Amplitud)

```python
def bfs(grafo, inicio, objetivo):
    """
    Busca el camino con menos pasos desde el inicio hasta el objetivo.
    """

    # La cola guarda los caminos que se van formando.
    # Al inicio, solo existe el camino que contiene el nodo inicial.
    cola = [[inicio]]

    # set() crea un conjunto vacío.
    # Aquí se guardan los nodos ya revisados para no visitarlos otra vez.
    visitados = set()

    # El ciclo continúa mientras existan caminos pendientes en la cola.
    while cola:

        # pop(0) saca el primer elemento de la lista.
        # Por eso BFS trabaja como una cola: primero en entrar, primero en salir.
        camino = cola.pop(0)

        # camino[-1] obtiene el último nodo del camino actual.
        nodo_actual = camino[-1]

        # Si el nodo actual es el objetivo, se devuelve el camino encontrado.
        if nodo_actual == objetivo:
            return camino

        # Solo se revisa el nodo si todavía no fue visitado.
        if nodo_actual not in visitados:

            # Agrega el nodo actual al conjunto de nodos visitados.
            visitados.add(nodo_actual)

            # Recorre los nodos conectados directamente con el nodo actual.
            for vecino in grafo[nodo_actual]:

                # Crea una copia del camino actual y agrega el vecino.
                nuevo_camino = camino + [vecino]

                # Agrega el nuevo camino al final de la cola.
                cola.append(nuevo_camino)

    # Se devuelve None si no se encontró ningún camino.
    return None


# Busca un camino desde Almacen hasta ZonaNorte.
resultado = bfs(mapa, "Almacen", "ZonaNorte")

# Muestra el camino encontrado.
print("Camino encontrado:", resultado)
```

#### 4.4. Vectores y matrices con NumPy

**Parte A:** Representar cuatro observaciones de un dataset simple (por ejemplo, sensores de temperatura y humedad) como un vector cada una, y juntarlas en una matriz:

```python
import numpy as np

# Cada vector representa una observación del dataset.
# En este caso, cada observación tiene temperatura y humedad.
obs1 = np.array([22.5, 60])
obs2 = np.array([19.0, 45])
obs3 = np.array([25.3, 70])
obs4 = np.array([21.0, 50])

# La matriz dataset reúne las cuatro observaciones.
# Cada fila es una observación y cada columna es una característica.
# La primera columna representa temperatura y la segunda humedad.
dataset = np.array([obs1, obs2, obs3, obs4])

print(dataset) # Se muestra la matriz completa de observaciones.

print("Forma de la matriz (filas, columnas):", dataset.shape) # shape muestra la cantidad de filas y columnas de la matriz.
```

**Parte B:** Acceder a una la y a una columna especíca de la matriz:

```python
# Se accede a la primera observación del dataset, dataset[0] representa la fila 0 de la matriz.
print("Primera observacion (fila 0):", dataset[0])

# Se accede a la primera característica de todas las observaciones, dataset[:, 0] representa todas las filas de la columna 0: temperaturas.
print("Todas las temperaturas (columna 0):", dataset[:, 0])
```

#### 4.5. Producto escalar y distancia euclidiana

**Parte A:** Calcular el producto escalar entre dos observaciones:

```python
producto = np.dot(obs1, obs2) # Se calcula el producto escalar entre dos observaciones
print("Producto escalar entre obs1 y obs2:", producto) # Se muestra el producto escalar calculadoo.ucto)
```

**Parte B:** Calcular la distancia euclidiana entre dos observaciones, primero con la fórmula manual
y luego con NumPy:

```python
# Se calcula la distancia euclidiana de forma manual.
# Primero se resta obs1 menos obs2, luego se eleva cada diferencia al cuadrado,se suman los resultados y finalmente se calcula la raíz cuadrada.
distancia_manual = np.sqrt(np.sum((obs1 - obs2) ** 2))

print("Distancia manual:", distancia_manual) # Se muestra la distancia manual.

distancia_numpy = np.linalg.norm(obs1 - obs2) # Se calcula la misma distancia euclidiana usando una función de NumPy.

print("Distancia con NumPy:", distancia_numpy) # Se muestra la distancia calculada con NumPy.
```

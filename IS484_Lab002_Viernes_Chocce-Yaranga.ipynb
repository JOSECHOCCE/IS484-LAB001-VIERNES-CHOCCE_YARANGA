#### 4.2. Representación de un problema mediante estados
**Parte A:** Completar la ficha del problema para el siguiente caso: un repartidor debe ir desde el Almacén hasta la Zona Norte, pudiendo pasar por distintos puntos intermedios de la ciudad.

| Elemento | Descripción |
| :--- | :--- |
| **Estado inicial** | Repartidor en el Almacén. |
| **Estado objetivo** | Repartidor en la Zona Norte. |
| **Acciones posibles** | Desplazarse hacia un punto directamente conectado con la ubicación actual (moverse a un nodo vecino). |
| **Costo de cada acción** | 1 paso (costo unitario por desplazamiento). |ento). |

**Parte B:**  Representar el mapa de la ciudad como un grafo en Python, usando un diccionario de listas (cada clave es un punto, cada valor es la lista de puntos conectados directamente):

```python
mapa = {
    "Almacen": ["Centro", "Terminal"],  # Desde el Almacén se puede ir a Centro o Terminal
    "Centro": ["Almacen", "ZonaNorte", "Mercado"],  # Centro conecta tres puntos
    "Terminal": ["Almacen", "Mercado"],  # Terminal conecta con Almacén y Mercado
    "Mercado": ["Centro", "Terminal", "ZonaNorte"],  # Mercado conecta con tres puntos
    "ZonaNorte": ["Centro", "Mercado"],  # Zona Norte puede regresar a Centro o Mercado
}

print(mapa)  # Muestra el diccionario que representa el grafo de la ciudad
```

#### 4.3. Implementación de BFS (Búsqueda en Amplitud)

```python
def bfs(grafo, inicio, objetivo):
    """
    Busca el camino con menos pasos entre inicio y objetivo usando BFS.
    BFS utiliza una cola: el primer camino que entra es el primero que sale.
    """

    cola = [[inicio]]  # La cola guarda caminos; inicia con el camino que contiene solo el nodo inicial
    visitados = set()  # El conjunto evita volver a procesar nodos que ya fueron revisados

    while cola:
        camino = cola.pop(0)  # pop(0) retira el primer elemento: comportamiento FIFO de una cola
        nodo_actual = camino[-1]  # El último nodo del camino es el nodo que se está explorando

        if nodo_actual == objetivo:
            return camino  # Si se llegó al objetivo, se devuelve el camino encontrado

        if nodo_actual not in visitados:
            visitados.add(nodo_actual)  # Se marca el nodo actual para no visitarlo nuevamente

            for vecino in grafo[nodo_actual]:
                nuevo_camino = camino + [vecino]  # Se crea una copia del camino agregando el vecino
                cola.append(nuevo_camino)  # El nuevo camino se coloca al final de la cola

    return None  # Se devuelve None si no existe un camino entre inicio y objetivo


resultado = bfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado:", resultado)
```

#### 4.4. Vectores y matrices con NumPy

**Parte A:** Representar cuatro observaciones de un dataset simple (por ejemplo, sensores de temperatura y humedad) como un vector cada una, y juntarlas en una matriz:

```python
import numpy as np  # NumPy permite trabajar con vectores y matrices numéricas

obs1 = np.array([22.5, 60])  # Primera observación: temperatura y humedad
obs2 = np.array([19.0, 45])  # Segunda observación: temperatura y humedad
obs3 = np.array([25.3, 70])  # Tercera observación: temperatura y humedad
obs4 = np.array([21.0, 50])  # Cuarta observación: temperatura y humedad

dataset = np.array([obs1, obs2, obs3, obs4])
# La matriz dataset reúne las cuatro observaciones: cada fila representa una observación y cada columna una característica

print(dataset)
print("Forma de la matriz (filas, columnas):", dataset.shape)
```

#### 4.5. Producto escalar y distancia euclidiana

**Parte A:** Calcular el producto escalar entre dos observaciones:

```python
producto = np.dot(obs1, obs2)
# np.dot calcula el producto escalar multiplicando los valores correspondientes de ambos vectores y sumando los resultados

print("Producto escalar entre obs1 y obs2:", producto)
```

**Parte B:** Calcular la distancia euclidiana entre dos observaciones, primero con la fórmula manual
y luego con NumPy:

```python
# Fórmula manual de la distancia euclidiana:
# resta los vectores, eleva las diferencias al cuadrado, las suma y calcula la raíz cuadrada
distancia_manual = np.sqrt(np.sum((obs1 - obs2) ** 2))
print("Distancia manual:", distancia_manual)

# np.linalg.norm calcula directamente la norma del vector diferencia, que en este caso representa la distancia euclidiana entre obs1 y obs2
distancia_numpy = np.linalg.norm(obs1 - obs2)
print("Distancia con NumPy:", distancia_numpy)
```

#### 5. Ejercicio Práctico de Laboratorio

**5.1. Planteamiento del problema:** Se quiere encontrar, dentro de un conjunto de observaciones ya conocidas, cuál es la más parecida a
una observación nueva  el mismo principio que usará el algoritmo KNN a partir de la Semana 7.

**5.2. Parte A:** Implementación de DFS

```python
def dfs(grafo, inicio, objetivo):
    """
    Busca un camino entre inicio y objetivo usando DFS.
    DFS utiliza una pila: el último camino que entra es el primero que sale.
    """

    pila = [[inicio]]  # La pila guarda caminos y comienza con el estado inicial
    visitados = set()  # Evita procesar varias veces el mismo nodo

    while pila:
        camino = pila.pop()  # pop() retira el último elemento: comportamiento LIFO de una pila
        nodo_actual = camino[-1]  # Se obtiene el último nodo del camino actual

        if nodo_actual == objetivo:
            return camino  # Si el nodo actual es el objetivo, se devuelve el camino

        if nodo_actual not in visitados:
            visitados.add(nodo_actual)  # Se registra el nodo como visitado

            for vecino in grafo[nodo_actual]:
                nuevo_camino = camino + [vecino]  # Se genera un camino nuevo agregando un vecino
                pila.append(nuevo_camino)  # El camino se agrega al final de la pila

    return None  # Se devuelve None si no se encuentra un camino


resultado_dfs = dfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado (DFS):", resultado_dfs)
```

**5.3. Parte B:**  Encontrar la observación más cercana

```python
nueva_observacion = np.array([20.5, 48])
# Esta nueva observación contiene temperatura = 20.5 y humedad = 48

distancias = []  # Lista donde se guardarán las distancias a cada observación del dataset

for i, obs in enumerate(dataset):
    # Se calcula la distancia euclidiana entre la nueva observación y cada fila del dataset
    d = np.linalg.norm(nueva_observacion - obs)

    distancias.append(d)  # Se guarda la distancia calculada para compararla después
    print(f"Distancia a la observacion {i}: {d}")

indice_mas_cercano = np.argmin(distancias)
# np.argmin devuelve la posición del valor más pequeño dentro de la lista de distancias

print("La observacion mas parecida es la numero", indice_mas_cercano)
```

#### 5.4. Parte C: Reflexión

La distancia euclidiana es una forma razonable de medir qué tan parecidas son dos observaciones porque calcula la separación directa entre sus valores numéricos. Si la distancia es pequeña, las observaciones tienen características similares; si es grande, son más diferentes.

Sin embargo, no siempre es la mejor opción. Puede dar resultados poco adecuados cuando las características tienen escalas muy distintas, por ejemplo, si una columna representa ingresos en soles y otra representa edad en años. En esos casos, primero se deberían normalizar los datos para que una característica no tenga más influencia que otra.
